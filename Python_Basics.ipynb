```python
## H5: Write a program to calculate and print the average of the three numbers. Round the result to two decimal places.

# Take three numbers as input from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

# Calculate the average
average = (num1 + num2 + num3) / 3

# Method 1: Using the round() function
rounded_avg = round(average, 2)
print("\nAverage (using round):", rounded_avg)
print(f"Average (formatted):   {average:.2f}")
```

```python
## H6: Create a simple calculator that performs basic arithmetic operations (addition, subtraction, multiplication, division) based on user input.


num1 = float(input("Enter the first number: "))
operator = input("Choose an operator (+, -, *, /): ")
num2 = float(input("Enter the second number: "))


if operator == "+":
    result = num1 + num2
    print(f"\nResult: {num1} + {num2} = {result:.2f}")

elif operator == "-":
    result = num1 - num2
    print(f"\nResult: {num1} - {num2} = {result:.2f}")

elif operator == "*":
    result = num1 * num2
    print(f"\nResult: {num1} * {num2} = {result:.2f}")

elif operator == "/":
    if num2 == 0:
        print("\nError: Cannot divide by zero.")
    else:
        result = num1 / num2
        print(f"\nResult: {num1} / {num2} = {result:.2f}")

else:
    print("\nError: Invalid operator. Please choose +, -, *, or /.")


```

```python
## H8: Create a program that converts temperatures between Fahrenheit and Celsius based on user input.


unit = input("Enter unit (C or F): ")
temp = float(input("Enter temperature: "))

if unit == "C":
    fah = (temp * 9/5) + 32
    print("In Fahrenheit:", fah)
elif unit == "F":
    cel = (temp - 32) * 5/9
    print("In Celsius:", cel)
else:
    print("Invalid unit entered.")

```

```python
# H10: Write a program that prompts the user to input a number N and prints all even numbers between 1 and N using a for loop.


# Prompt the user for a number N
N = int(input("Enter a number N: "))

print(f"Even numbers between 1 and {N}:")

# Loop from 1 up to N (inclusive)
for i in range(1, N + 1):
    if i % 2 == 0:
        print(i)

```

```python
# H11: Write a program that prompts the user to input a number N and calculates the sum of all numbers from 1 to N that are divisible by 3 or 5.

# Prompt the user for a number N
N = int(input("Enter a number N: "))

total_sum = 0


for i in range(1, N + 1):
    if i % 3 == 0 or i % 5 == 0:
        total_sum = total_sum + i

print(f"The sum of numbers divisible by 3 or 5 up to {N} is: {total_sum}")




```

```python
# H12: Write a program that prompts the user to input a number n and prints the multiplication table of n (1 to 10) using a for loop.

# Prompt the user for a number n
n = int(input("Enter a number: "))

print(f"\nMultiplication Table of {n}:")


for i in range(1, 11):
    result = n * i
    print(f"{n} x {i} = {result}")


```
