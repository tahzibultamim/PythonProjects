```python
#H1 Write a function to calculate the sum of numbers from 1 to N.


def sum_to_n(n: int) -> int:
    """Calculate the sum of numbers from 1 to n."""
    if n < 1:
        return 0
    return n * (n + 1) // 2


# Get input from the user
try:
    user_n = int(input("Enter a positive integer (N): "))
    if user_n < 1:
        print("Please enter a number greater than 0.")
    else:
        result = sum_to_n(user_n)
        print(f"The sum of numbers from 1 to {user_n} is: {result}")
except ValueError:
    print("Invalid input! Please enter a valid integer.")
```

```python
#H2 Write a function that returns the square of a number.
def square(n: float) -> float:
    """Return the square of a number."""
    return n ** 2


# Example usage with user input
try:
    num = float(input("Enter a number: "))
    print(f"The square of {num} is: {square(num)}")
except ValueError:
    print("Invalid input! Please enter a valid number.")
```

```python
#H3 Write a function that checks whether a number is even/odd.

def check_even_odd(n: int) -> str:
    """Return whether an integer is 'Even' or 'Odd'."""
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"


try:
    num = int(input("Enter an integer: "))
    result = check_even_odd(num)
    print(f"{num} is {result}.")
except ValueError:
    print("Invalid input! Please enter a valid integer.")
```

```python
#H4 Write a function that returns the larger number from two given numbers.

def get_larger(a: float, b: float) -> float:
    """Return the larger of two numbers."""
    if a > b:
        return a
    return b

try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    larger = get_larger(num1, num2)
    print(f"The larger number is: {larger}")
except ValueError:
    print("Invalid input! Please enter valid numbers.")
```

```python
## String Functions

# Initial string
s = "Skill Morph"

# 1. upper()
print(s.upper())  # Converts all characters to uppercase
# 2. lower()
print(s.lower())  # Converts all characters to lowercase
# 3. partition()
print(s.partition(' '))  # Splits the string at the first occurrence of ' '
# 4. replace()
print(s.replace('Morph', 'Master'))  # Replaces 'Morph' with 'Master'
# 5. find()
print(s.find('Morph'))  # Returns the index of 'Morph' in the string, or -1 if not found
# 6. rstrip()
print(s.rstrip())  # Removes trailing spaces from the string
# 7. split()
print(s.split())  # Splits the string into a list at spaces
# 8. startswith()
print(s.startswith('Skill'))  # Checks if the string starts with 'Skill'
# 9. isnumeric()
print(s.isnumeric())  # Checks if all characters in the string are numeric (returns False)
# 10. index()
print(s.index('Morph'))  # Returns the index of the first occurrence of 'Morph'
```
